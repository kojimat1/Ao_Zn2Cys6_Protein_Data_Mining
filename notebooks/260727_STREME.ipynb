```python
CONDA = "/content/miniforge3/bin/conda"

!{CONDA} config --system --remove-key channels 2>/dev/null || true
!{CONDA} config --system --add channels conda-forge
!{CONDA} config --system --add channels bioconda
!{CONDA} config --system --set channel_priority strict

!{CONDA} config --show channels
!{CONDA} config --show channel_priority
```

```python
!{CONDA} create \
  -y \
  -n meme_env \
  --override-channels \
  -c conda-forge \
  -c bioconda \
  python=3.10 \
  meme=5.5.9
```

```python
!{CONDA} run -n meme_env streme --version
```

```python
!{CONDA} run -n meme_env which streme
```

```python
import shutil
from google.colab import files
from pathlib import Path
```

STREME DBD

```python
!{CONDA} run -n meme_env streme \
    --protein \
    --p "/content/aor0-5_SM00066_domains_2.faa" \
    --o "/content/260727_streme_DBD_output" \
    --minw 12 \
    --maxw 14
```

```python
# ===== Create a ZIP archive of the output directory =====
output_path = Path("/content/260727_streme_DBD_output")

zip_path = shutil.make_archive(
    "260727_streme_DBD_output",
    'zip',
    root_dir=output_path
)

print(f"ZIP file: {zip_path}")

# ===== Download =====
files.download(zip_path)
```
